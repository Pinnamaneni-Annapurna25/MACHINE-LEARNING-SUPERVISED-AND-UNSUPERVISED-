## B1. Customer Segmentation Script

The original code works, but it has repeated code for scaling,calculating silhouette scores, and creating the cluster profile.I will refactor it to make the code cleaner, reusable, and reproducible.
The refactored version will:
- Use StandardScaler instead of manual scaling.
- Store the features in a list.
- Use a loop to test different values of k.
- Use random_state for reproducible results.
- Automatically select the k with the highest silhouette score.
- Create the cluster profile using one groupby() operation.

````python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score


def customer_segmentation(
    file_path,
    features,
    k_range=range(2, 6),
    random_state=42
):
    # Load data
    df = pd.read_csv('/Users/pinnamanenisaisubhash/Downloads/customer_data.csv')

    # Select features
    X = df[features]

    # Scale features using scikit-learn
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Calculate silhouette score for each k
    silhouette_scores = {}

    for k in k_range:
        model = KMeans(
            n_clusters=k,
            n_init=10,
            random_state=random_state
        )

        labels = model.fit_predict(X_scaled)

        silhouette_scores[k] = silhouette_score(
            X_scaled,
            labels
        )

    # Select k with the highest silhouette score
    best_k = max(
        silhouette_scores,
        key=silhouette_scores.get
    )

    # Train final K-Means model
    final_model = KMeans(
        n_clusters=best_k,
        n_init=10,
        random_state=random_state
    )

    df["cluster"] = final_model.fit_predict(X_scaled)

    # Create cluster profile in one step
    profile = df.groupby("cluster")[features].mean()

    return df, profile, silhouette_scores, best_k
````

# B2.  Dimensionality reduction helper   

## B2(a) Problems in the Original Function

The original function has the following problems:

1. **Data leakage:** The test data is used to calculate its own mean and
   standard deviation. The scaling values should be calculated only
   from the training data.

2. **Different scaling:** The training and test data are standardised
   using different means and standard deviations. This makes the train
   and test data inconsistent.

3. **Inefficient PCA:** PCA is fitted repeatedly inside the loop for
   every possible number of components.

4. **Not configurable:** The variance target is fixed at 90% instead
   of being provided as a parameter.

5. **Possible failure:** best can remain None if the required
   variance target is not reached.

# B2(b) Rewrite the function

````python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA


def reduce_dimensions(train, test, variance_target=0.90):
    """
    Standardise train and test data using training statistics,
    then reduce the data using PCA while retaining the required
    amount of variance.
    """

    # Standardise using only training data
    scaler = StandardScaler()

    train_scaled = scaler.fit_transform(train)
    test_scaled = scaler.transform(test)

    # Fit PCA once to find explained variance
    pca_full = PCA()
    pca_full.fit(train_scaled)

    # Find the smallest number of components
    # that explains the required variance
    cumulative_variance = np.cumsum(
        pca_full.explained_variance_ratio_
    )

    best = np.searchsorted(
        cumulative_variance,
        variance_target
    ) + 1

    # Fit PCA with the selected number of components
    pca = PCA(n_components=best)

    train_reduced = pca.fit_transform(train_scaled)
    test_reduced = pca.transform(test_scaled)

    return train_reduced, test_reduced
````

## B2(c) Change Log

- Used StandardScaler and fitted it only on the training data to
  prevent data leakage.
- Fitted PCA once to calculate cumulative explained variance instead
  of fitting it repeatedly.
- Added variance_target as a parameter so the required variance can
  be changed.
- Used the same fitted scaler and PCA transformation process for both
  training and test data.

````python
# Example data
rng = np.random.RandomState(42)

train = rng.rand(100, 10)
test = rng.rand(20, 10)

train_reduced, test_reduced = reduce_dimensions(
    train,
    test,
    variance_target=0.90
)

print("Original train shape:", train.shape)
print("Reduced train shape:", train_reduced.shape)

print("Original test shape:", test.shape)
print("Reduced test shape:", test_reduced.shape)
````

### D2(a)More Components Requested Than the Data Allows
PREDICTION:
When requesting more PCA components than possible (n_components=10 for only 5 samples × 20 features), PCA will be limited by the number of samples, not features. The maximum components possible is min(n_samples, n_features) - 1 = min(5, 20) - 1 = 4.

````python
import numpy as np
from sklearn.decomposition import PCA

X = np.random.RandomState(0).rand(5, 20)  # 5 samples, 20 features
pca = PCA(n_components=min(10, X.shape[0] - 1)).fit(X)
print(pca.n_components_)  # Output: 4
print(pca.explained_variance_ratio_.shape)  # Output: (4,)
````

## Why This Happens:
PCA decomposes the covariance matrix, which has rank at most min(n_samples - 1, n_features)
With 5 samples, only 4 independent components can exist (not 20)
scikit-learn silently caps the components; it does not raise an error

````python
# Robust Handling of PCA Component Selection
def safe_pca(X, target_components):
    max_allowed = min(X.shape[0] - 1, X.shape[1])
    actual_components = min(target_components, max_allowed)
    if actual_components < target_components:
        print(f"⚠️  Capped n_components from {target_components} to {actual_components}")
    return PCA(n_components=actual_components).fit(X)
````

# Rule for Maximum Components:
max_components = min(n_samples - 1, n_features)

- `preferred_channel` — captures the customer's preferred interaction channel.

I will exclude `customer_id` because it is only an identifier and has no meaningful behavioural information.

### Preprocessing Decisions

- Missing `monthly_spend` values will be replaced using the median. Median imputation is less affected by the highly variable spending values.
- `monthly_spend` will receive a `log1p` transformation because spending is strongly right-skewed.
- Numerical features will be standardised so that variables measured on different scales do not dominate clustering.
- `preferred_channel` will be converted using one-hot encoding.
- PCA will reduce the processed feature space while retaining 90% of the variance.
- K-Means will then be applied to the PCA representation.

````python
import numpy as np
import pandas as pd

def make_customers(n=1500, seed=42):
    rng = np.random.default_rng(seed)

    seg = rng.choice(
        4,
        size=n,
        p=[0.40, 0.30, 0.20, 0.10]
    )

    # (monthly_spend, orders_per_month, return_rate)
    spec = {
        0: (300, 6, 0.30),
        1: (900, 14, 0.10),
        2: (2500, 30, 0.05),
        3: (6000, 12, 0.02)
    }

    spend = np.array([
        max(5, rng.normal(spec[s][0], spec[s][0] * 0.25))
        for s in seg
    ])

    orders = np.array([
        max(1, rng.poisson(spec[s][1]))
        for s in seg
    ])

    ret = np.array([
        min(1, max(0, rng.normal(spec[s][2], 0.05)))
        for s in seg
    ])

    df = pd.DataFrame({
        "customer_id": np.arange(1, n + 1),
        "monthly_spend": spend.round(2),
        "orders_per_month": orders,
        "return_rate": ret.round(3),
        "tenure_months": rng.integers(1, 60, n),
        "preferred_channel": rng.choice(
            ["web", "app", "store"],
            n,
            p=[0.45, 0.40, 0.15]
        )
    })

    df.loc[
        rng.choice(n, 60, replace=False),
        "monthly_spend"
    ] = np.nan

    return df


df = make_customers()

print("Shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())

df.head()
````

# E2. Pipeline Implementation 

A single scikit-learn pipeline is used to preprocess the customer data, reduce
dimensionality using PCA, and perform K-Means clustering.

The pipeline includes:
- Median imputation for missing monthly_spend values
- Log transformation of monthly_spend
- Standard scaling of numerical features
- One-hot encoding of preferred_channel
- PCA retaining 90% of the variance
- K-Means clustering

The random_state is fixed at 42 to make the results reproducible.


````python
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import FunctionTransformer
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans


# Features
spend_feature = ["monthly_spend"]

numeric_features = [
    "orders_per_month",
    "return_rate",
    "tenure_months"
]

categorical_features = [
    "preferred_channel"
]


# Pipeline for monthly spend
spend_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("log_transform", FunctionTransformer(np.log1p)),
    ("scaler", StandardScaler())
])


# Pipeline for other numerical features
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])


# Pipeline for categorical feature
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])


# Combine all preprocessing steps
preprocessor = ColumnTransformer([
    ("spend", spend_pipeline, spend_feature),
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])


# Complete end-to-end pipeline
segmentation_pipeline = Pipeline([
    ("preprocessing", preprocessor),

    ("pca", PCA(
        n_components=0.90,
        svd_solver="full"
    )),

    ("clustering", KMeans(
        n_clusters=2,
        n_init=20,
        random_state=42
    ))
])


# Fit the complete pipeline
segmentation_pipeline.fit(df)


print("Pipeline fitted successfully.")
````

### E2 Explanation

The complete segmentation workflow is implemented as one reproducible scikit-learn Pipeline. The ColumnTransformer applies different preprocessing to numerical and categorical features. Missing numerical values are imputed,monthly_spend is log-transformed because of its skew, numerical features are standardised, and preferred_channel is one-hot encoded.

The processed data is then passed to PCA, which retains 90% of the variance.Finally, K-Means assigns each customer to a segment. The random_state is set to 42 and n_init is fixed at 20 for reproducibility.

Because preprocessing, PCA and clustering are all stored in the same Pipeline,the same transformations are automatically applied when predicting newcustomers. Therefore, new customers can be assigned to an existing segmentusing segmentation_pipeline.predict().
