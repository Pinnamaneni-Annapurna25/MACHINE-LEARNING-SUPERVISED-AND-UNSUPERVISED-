# K-MEANS CLUSTERING

```python

from sklearn.cluster import KMeans


inertia = []

K_range = range(2, 11)

for k in K_range:
    
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    K_range,
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.grid()
plt.show()
```

```python
from sklearn.metrics import silhouette_score
silhouette_scores = []

for k in range(2, 11):
    
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_scaled)
    
    score = silhouette_score(
        X_scaled,
        labels
    )
    
    silhouette_scores.append(score)
    
    print(
        "K =", k,
        "Silhouette Score =", round(score, 4)
    )
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.grid()
plt.show()
```

```python
k_values = list(K_range)
inertias = inertia

results = pd.DataFrame({
    "K": k_values,
    "Inertia_WCSS": inertias,
    "Silhouette_Score": silhouette_scores
})

results
```

```python
kmeans_final = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans_final.fit_predict(X_scaled)
```

```python
df['KMeans_Cluster'] = kmeans_labels
```

```python
df['KMeans_Cluster'].value_counts().sort_index()
```

```python
kmeans_silhouette = silhouette_score(
    X_scaled,
    kmeans_labels
)

print("K-Means Silhouette Score:",
      kmeans_silhouette)
```

# DBSCAN

```python
# Select features for DBSCAN

dbscan_features = [
    "Annual_Income",
    "Total_Amount_Spent"
]

X_dbscan = df[dbscan_features].copy()

X_dbscan.head()
```

```python
# Train DBSCAN

from sklearn.cluster import DBSCAN

dbscan_model = DBSCAN(
    eps=0.3,
    min_samples=10
)

# Standardize the DBSCAN features before clustering
X_dbscan_scaled = StandardScaler().fit_transform(X_dbscan)

dbscan_labels = dbscan_model.fit_predict(X_dbscan_scaled)
```

```python
# DBSCAN Cluster Distribution

df_dbscan = df.copy()
df_dbscan["DBSCAN_Cluster"] = dbscan_labels

df_dbscan["DBSCAN_Cluster"].value_counts().sort_index()
```

```python
# Test different eps values

eps_values = [0.10, 0.15, 0.20, 0.25, 0.30]

for eps in eps_values:
    model = DBSCAN(
        eps=eps,
        min_samples=10
    )
    
    labels = model.fit_predict(X_dbscan_scaled)
    
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    noise_count = (labels == -1).sum()
    
    print(
        f"eps={eps:.2f} | "
        f"Clusters={n_clusters} | "
        f"Noise={noise_count} "
        f"({noise_count / len(labels) * 100:.2f}%)"
    )
```

```python
# Final DBSCAN Model

dbscan_final = DBSCAN(
    eps=0.15,
    min_samples=10
)

dbscan_labels = dbscan_final.fit_predict(X_dbscan_scaled)

df_dbscan["DBSCAN_Cluster"] = dbscan_labels
```

```python
# Final DBSCAN Cluster Distribution

df_dbscan["DBSCAN_Cluster"].value_counts().sort_index()
```

```python
# Cell 8 — DBSCAN Customer Segmentation

plt.figure(figsize=(10, 6))

plt.scatter(
    df_dbscan["Annual_Income"],
    df_dbscan["Total_Amount_Spent"],
    c=df_dbscan["DBSCAN_Cluster"],
    s=30,
    alpha=0.6
)

plt.title("DBSCAN Customer Segmentation")
plt.xlabel("Annual Income")
plt.ylabel("Total Amount Spent")
plt.colorbar(label="DBSCAN Cluster")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# DBSCAN Customer Profile

dbscan_profile = (
    df_dbscan
    .groupby("DBSCAN_Cluster")[[
        "Annual_Income",
        "Total_Amount_Spent"
    ]]
    .mean()
    .round(2)
)

dbscan_profile
```

# HIERARCHICAL CLUSTERING

```python
from sklearn.cluster import AgglomerativeClustering

sample_size = 1000

X_sample = X_scaled[
    np.random.RandomState(42).choice(
        X_scaled.shape[0],
        sample_size,
        replace=False
    )
]
sample_size = 1000

X_sample = X_scaled[
    np.random.RandomState(42).choice(
        X_scaled.shape[0],
        sample_size,
        replace=False
    )
]
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram

linked = linkage(
    X_sample,
    method='ward'
)

plt.figure(figsize=(15,7))

dendrogram(
    linked,
    truncate_mode='lastp',
    p=30
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Cluster")
plt.ylabel("Distance")

plt.show()
```

```python
hierarchical = AgglomerativeClustering(
    n_clusters=4,
    linkage='ward'
)

hierarchical_labels = hierarchical.fit_predict(
    X_scaled
)
```
